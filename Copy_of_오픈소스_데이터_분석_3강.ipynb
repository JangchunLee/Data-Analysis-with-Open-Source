```python
def get_dataset_stats(values):
  return min(values), max(values), sum(values) / len(values)

data = [12, 8, 21, 17, 5, 32, 14]

# 함수값 return 언패킹
minimum, maximum, average = get_dataset_stats(data)
print(f'최솟값: {minimum}, 최댓값: {maximum}, 평균: {average:.2f}')
```

```python
student_scores = [
    ('김철수', 85, 92, 78),
    ('이영희', 92, 88, 95),
    ('박지민', 75, 83, 90)
]

# 반복문에서 언패킹 사용
for name, database, python, cloud in student_scores:
  average = (database + python + cloud) / 3
  print(f'{name}의 평균 점수: {average:.1f}')
```

```python
monthly_sales = [1200, 1350, 1420, 1500, 1300, 1580, 1620, 1700, 1800, 1850, 1900, 2000]

# 1월, 2월 및 나머지 월 매출액
first, second, *remaining = monthly_sales
print(f'1월 판매액: {first}')
print(f'2월 판매액: {second}')
print(f'나머지 월 판매액: {remaining}')

# 1월, 12월 및 나머지 월 매출액
first, *middle, eleventh, last = monthly_sales
print(f'첫 달(1월) 판매액: {first}')
print(f'마지막 달(12월) 판매액: {last}')
print(f'11월:{eleventh}')
print(f'중간 달(2~11월)의 판매액: {middle}')
```

## 3-3 함수형 프로그래밍

```python
employees = {
    '이지혜': 3000000,
    '구민준': 5000000,
    '방서연': 4100000
}

# 함수 정의: salary * 1.1
adjust_salary = lambda salary: salary * 1.1
# 급여 인상
updated_salaries = {name: adjust_salary(salary) for name, salary in employees.items()}
print(updated_salaries)
```

```python
temperature_celsius = [25.6, 27.8, 30.5, 22.3, 28.9, 31.2, 24.7]

# 함수 정의: 섭씨 -> 화씨
cel_to_fahrenheit = lambda celsius: (celsius*9/5) +32
# 온도 변환
temp_fahrenheit = {cel_to_fahrenheit(celsius) for celsius in temperature_celsius}
temperature_farenheit = list(map(cel_to_fahrenheit, temperature_celsius))

print('섭씨 온도 데이터:', temperature_celsius)
print('화씨 온도로 변환:', temperature_fahrenheit)
```

```python
words = ['python', 'data', 'analysis']

# 함수 정의 및 대문자화
capitalize = lambda word: word.upper()
uppercase_words = list(map(capitalize, words))
print(uppercase_words)
```

```python
employees = [
    {'name': '김지원', 'salary': 3700000},
    {'name': '박민준', 'salary': 7000000},
    {'name': '이서연', 'salary': 4400000},
    {'name': '정현우', 'salary': 3500000}
]

# salary * 1.1
adjust_salary = lambda salary: salary * 1.1
updated_salaries = [ adjust_salary(emp['salary']) for emp in employees ]
for emp in employees:
    print(emp['name'])
updated_employees = [
    {'name': emp['name'], 'salary': adjust_salary(emp['salary'])}
    for emp in employees
]
print('\n급여 인상 후:')
for emp in updated_employees:
    print(f'{emp['name']}: {int(emp['salary']):,}원')
```

```python
osda_students = [
    {'name': '김지원', 'score': 85},
    {'name': '이민준', 'score': 65},
    {'name': '박서연', 'score': 90},
    {'name': '정현우', 'score': 55},
    {'name': '최예은', 'score': 78}
]

# 70점 이상 학생 목록 생성
passOrFail = lambda score: score >= 70

passed_students = [
    student
    for student in osda_students
    if passOrFail(student['score'])
]

passed_students = list(filter(lambda student: passOrFail(student['score']), osda_students))
passed_students = list(filter(lambda student: student['score'] >= 70, osda_students))

for student in passed_students:
    print(f'이름: {student["name"]}, 점수: {student["score"]}')
```
